```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from scipy.stats import chi2_contingency
import pickle
import warnings

from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.metrics import mean_squared_error, r2_score

warnings.filterwarnings('ignore')

# ---------------------------------------------------------------------------
# 1. Load data
# ---------------------------------------------------------------------------
local_filename = r"C:\Users\HP\Downloads\cardekho_dataset.csv"
df = pd.read_csv(local_filename, index_col=0)

print("\nTop 5 rows of the dataset:")
print("-" * 100)
print(df.head())

print("\nDataset info:")
print("-" * 100)
df.info()
print("Size of dataframe is:", df.shape)

numerical_features = [
    feature for feature in df.columns
    if df[feature].dtype != 'object' and df[feature].dtype != 'string'
]
print("-" * 100)
print("There are {} numerical features: {}".format(len(numerical_features), numerical_features))

categorical_features = [
    feature for feature in df.columns
    if df[feature].dtype == 'string' or df[feature].dtype == 'object'
]
print("-" * 100)
print("There are {} categorical features: {}".format(len(categorical_features), categorical_features))

print("Missing values:")
print("-" * 100)
print(df.isnull().sum())

print("Percentage of missing values:")
print("-" * 100)
missing_value = df.isnull().sum() / len(df) * 100
print(missing_value)

print("Statistical summary of numerical features:")
print("-" * 100)
print(df.describe())

print("Statistical summary of categorical features:")
print("-" * 100)
print(df.describe(include='object'))

# ---------------------------------------------------------------------------
# 2. Unique value inspection
# ---------------------------------------------------------------------------
print("Unique features")
print("-" * 100)
print("'car_name' has {} unique categories:\n{}\n".format(df['car_name'].nunique(), df['car_name'].unique()))
print("-" * 100)
print("'brand' has {} unique categories:\n{}\n".format(df['brand'].nunique(), df['brand'].unique()))
print("-" * 100)
print("'model' has {} unique categories:\n{}\n".format(df['model'].nunique(), df['model'].unique()))
print("-" * 100)
print("'seller_type' has {} unique categories:\n{}\n".format(df['seller_type'].nunique(), df['seller_type'].unique()))
print("-" * 100)
print("'fuel_type' has {} unique categories:\n{}\n".format(df['fuel_type'].nunique(), df['fuel_type'].unique()))
print("-" * 100)
print("'transmission_type' has {} unique categories:\n{}\n".format(
    df['transmission_type'].nunique(), df['transmission_type'].unique()))
print("-" * 100)

print("Percentage of categorical feature:")
print("-" * 100)
for col in categorical_features:
    print(df[col].value_counts(normalize=True) * 100)
    print("-" * 100)

# ---------------------------------------------------------------------------
# 3. Univariate analysis
# ---------------------------------------------------------------------------
plt.figure(figsize=(15, 15))
plt.suptitle("Univariate analysis of numerical features:", fontsize=20, fontweight='bold', alpha=0.8)
for i in range(len(numerical_features)):
    plt.subplot(3, 3, i + 1)
    sns.kdeplot(x=df[numerical_features[i]], fill=True, color="b")
    plt.xlabel(numerical_features[i])
plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 15))
plt.suptitle("Univariate analysis of categorical features:", fontsize=20, fontweight='bold', alpha=0.8)
cat_1 = ["seller_type", "fuel_type", "brand", "transmission_type"]
for i in range(len(cat_1)):
    plt.subplot(2, 2, i + 1)
    sns.countplot(x=df[cat_1[i]])
    plt.xlabel(cat_1[i])
    plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

continuous_features = [feature for feature in numerical_features if df[feature].nunique() >= 10]
print("Continuous features:", continuous_features)

plt.figure(figsize=(15, 20))
for i in range(len(continuous_features)):
    plt.subplot(8, 2, i + 1)
    sns.scatterplot(data=df, x="selling_price", y=continuous_features[i], color="b")
    plt.xlim(0, 25000000)
plt.tight_layout()
plt.show()

# ---------------------------------------------------------------------------
# 4. Correlation heatmap
# ---------------------------------------------------------------------------
print(df[numerical_features].corr())
plt.figure(figsize=(15, 10))
sns.heatmap(data=df[numerical_features].corr(), annot=True, cmap="plasma", vmin=-1, vmax=1, linewidths=0.5)
plt.show()

# ---------------------------------------------------------------------------
# 5. Chi-square association test (selling_price binned first)
# ---------------------------------------------------------------------------
price_bins = pd.qcut(df['selling_price'], q=5, duplicates='drop')
chi2_test = []
for feature in categorical_features:
    p_value = chi2_contingency(pd.crosstab(price_bins, df[feature]))[1]
    chi2_test.append("Reject Null Hypothesis" if p_value < 0.05 else "Fail to Reject Null Hypothesis")

test_result = pd.DataFrame(data=[categorical_features, chi2_test]).T
test_result.columns = ["Categorical Features", "Hypothesis Result"]
print(test_result)

# ---------------------------------------------------------------------------
# 6. Target / feature exploration plots
# ---------------------------------------------------------------------------
plt.subplots(figsize=(14, 7))
sns.histplot(df.selling_price, bins=200, kde=True, color='b')
plt.title("Selling Price Distribution", weight="bold", fontsize=20, pad=20)
plt.ylabel("Count", weight="bold", fontsize=12)
plt.xlabel("Selling price", weight="bold", fontsize=12)
plt.xlim(0, 3000000)
plt.show()

print('-' * 50)
print('Top 10 Sold Cars on CarDekho Website')
print('-' * 50)
print(df.car_name.value_counts()[0:10])

plt.subplots(figsize=(14, 7))
sns.countplot(x="car_name", data=df, ec="black", palette="Set1", order=df['car_name'].value_counts().index)
plt.title("Top 10 Most Sold Car", weight="bold", fontsize=20, pad=20)
plt.ylabel("Count", weight="bold", fontsize=20)
plt.xlabel("Car Name", weight="bold", fontsize=16)
plt.xticks(rotation=45)
plt.xlim(-1, 10.5)
plt.show()

i20 = df[df['car_name'] == 'Hyundai i20']['selling_price'].mean()
print(f'The mean price of Hyundai i20 is {i20:.2f} Rupees')

print('-' * 50)
print('Top 10 Most Sold Car Brands')
print('-' * 50)
print(df.brand.value_counts()[0:10])

car = df.groupby('car_name').selling_price.max().to_frame().sort_values('selling_price', ascending=False)[0:10]
print('-' * 50)
print('Top 10 Costlier Cars on CarDekho Website')
print('-' * 50)
print(car)

plt.subplots(figsize=(14, 7))
sns.barplot(x=car.index, y=car.selling_price, ec="black", palette="Set1")
plt.title("Car Name vs Highest Selling Price", weight="bold", fontsize=20, pad=20)
plt.ylabel("Selling Price", weight="bold", fontsize=15)
plt.xlabel("Car Name", weight="bold", fontsize=16)
plt.xticks(rotation=90)
plt.show()

mileage_car = df.groupby('car_name')['mileage'].mean().sort_values(ascending=False).head()
print('-' * 50)
print('Most Mileage Car on CarDekho Website')
print('-' * 50)
print(mileage_car.to_frame())

plt.subplots(figsize=(14, 7))
sns.barplot(x=mileage_car.index, y=mileage_car.values, ec="black", palette="Set1")
plt.title("Car Name vs Mileage", weight="bold", fontsize=20, pad=20)
plt.ylabel("Mileage in Kmpl", weight="bold", fontsize=15)
plt.xlabel("Car Name", weight="bold", fontsize=12)
plt.ylim(0, 27)
plt.xticks(rotation=45)
plt.show()

plt.subplots(figsize=(14, 7))
sns.scatterplot(x="km_driven", y='selling_price', data=df, ec="white", color='b', hue='fuel_type')
plt.title("Kilometer Driven vs Selling Price", weight="bold", fontsize=20, pad=20)
plt.ylabel("Selling Price", weight="bold", fontsize=20)
plt.xlim(-10000, 800000)
plt.ylim(-10000, 10000000)
plt.xlabel("Kilometer driven", weight="bold", fontsize=16)
plt.show()

fuel = df.groupby('fuel_type')['selling_price'].median().sort_values(ascending=False)
print(fuel.to_frame())

plt.subplots(figsize=(14, 7))
sns.barplot(x=fuel.index, y=fuel.values, ec="black", palette="Set2_r")
plt.title("Fuel Type vs Selling Price", weight="bold", fontsize=20, pad=20)
plt.ylabel("Median Selling Price", weight="bold", fontsize=15)
plt.xlabel("Fuel Type", weight="bold", fontsize=12)
plt.show()

plt.subplots(figsize=(14, 7))
sns.countplot(x=df.fuel_type, ec="black", palette="Set2_r")
plt.title("Fuel Type Count", weight="bold", fontsize=20, pad=20)
plt.ylabel("Count", weight="bold", fontsize=15)
plt.xlabel("Fuel Type", weight="bold", fontsize=12)
plt.show()

fuel_mileage = df.groupby('fuel_type')['mileage'].mean().sort_values(ascending=False)
print(fuel_mileage.to_frame())

plt.subplots(figsize=(14, 7))
sns.boxplot(x='fuel_type', y='mileage', data=df, palette="Set1_r")
plt.title("Fuel Type vs Mileage", weight="bold", fontsize=20, pad=20)
plt.ylabel("Mileage in Kmpl", weight="bold", fontsize=15)
plt.xlabel("Fuel Type", weight="bold", fontsize=12)
plt.show()

plt.subplots(figsize=(14, 7))
sns.scatterplot(x="mileage", y='selling_price', data=df, ec="white", color='b', hue='fuel_type')
plt.title("Mileage vs Selling Price", weight="bold", fontsize=20, pad=20)
plt.ylabel("Selling Price", weight="bold", fontsize=20)
plt.ylim(-10000, 100000)
plt.xlabel("Mileage", weight="bold", fontsize=16)
plt.show()

plt.subplots(figsize=(14, 7))
sns.histplot(x=df.mileage, ec="black", color='g', kde=True)
plt.title("Mileage Distribution", weight="bold", fontsize=20, pad=20)
plt.ylabel("Count", weight="bold", fontsize=15)
plt.xlabel("Mileage", weight="bold", fontsize=12)
plt.show()

plt.subplots(figsize=(20, 10))
sns.lineplot(x='vehicle_age', y='selling_price', data=df, color='b')
plt.ylim(0, 2500000)
plt.show()

print(df.groupby('vehicle_age')['mileage'].mean().sort_values(ascending=False))

plt.subplots(figsize=(14, 7))
sns.boxplot(x=df.vehicle_age, y=df.mileage, palette="Set1")
plt.title("Vehicle Age vs Mileage", weight="bold", fontsize=20, pad=20)
plt.ylabel("Mileage", weight="bold", fontsize=20)
plt.xlabel("Vehicle Age in Years", weight="bold", fontsize=16)
plt.show()

oldest = df.groupby('car_name')['vehicle_age'].max().sort_values(ascending=False).head(10)
print(oldest.to_frame())

plt.subplots(figsize=(14, 7))
sns.countplot(x='seller_type', data=df, palette="rocket_r")
plt.title("Seller Type Count", weight="bold", fontsize=20, pad=20)
plt.ylabel("Count", weight="bold", fontsize=15)
plt.xlabel("Seller Type", weight="bold", fontsize=12)
plt.show()

dealer = df.groupby('seller_type')['selling_price'].median().sort_values(ascending=False)
print(dealer.to_frame())

plt.subplots(figsize=(14, 7))
sns.barplot(x=dealer.index, y=dealer.values, ec="black", palette="Set2_r")
plt.title("Seller Type vs Selling Price", weight="bold", fontsize=20, pad=20)
plt.ylabel("Median Selling Price", weight="bold", fontsize=15)
plt.xlabel("Seller Type", weight="bold", fontsize=12)
plt.show()

# ---------------------------------------------------------------------------
# 7. Feature engineering
# ---------------------------------------------------------------------------
df_copy = df.copy()
df_copy.drop(['car_name', 'brand', 'model'], axis=1, inplace=True)
print(df_copy.head())

df_encoded = pd.get_dummies(df_copy, columns=['fuel_type', 'transmission_type', 'seller_type'], drop_first=True)

# Safety net: encode any leftover text columns instead of letting them hit the model raw
remaining_object_cols = df_encoded.select_dtypes(include='object').columns.tolist()
if remaining_object_cols:
    print("Encoding additional categorical columns found:", remaining_object_cols)
    df_encoded = pd.get_dummies(df_encoded, columns=remaining_object_cols, drop_first=True)

print(df_encoded.head())

X = df_encoded.drop('selling_price', axis=1)
Y = df_encoded['selling_price']
print("X shape:", X.shape)
print("Y shape:", Y.shape)

importance_model = ExtraTreesRegressor(random_state=3)
importance_model.fit(X, Y)
feat_importances = pd.Series(importance_model.feature_importances_, index=X.columns)
feat_importances.nlargest(10).plot(kind='barh')
plt.title("Top 10 Feature Importances", weight="bold", fontsize=16)
plt.tight_layout()
plt.show()

# ---------------------------------------------------------------------------
# 8. Train / test split and baseline model comparison
# ---------------------------------------------------------------------------
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3)
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

models = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(),
    'Lasso': Lasso(),
    'SVR': SVR(),
    'Decision Tree': DecisionTreeRegressor(random_state=3),
    'Random Forest': RandomForestRegressor(random_state=3),
    'Extra Trees': ExtraTreesRegressor(random_state=3)
}

results = []
for name, model in models.items():
    model.fit(X_train, Y_train)
    Y_pred = model.predict(X_test)
    mse = mean_squared_error(Y_test, Y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(Y_test, Y_pred)
    results.append({'Model': name, 'MSE': mse, 'RMSE': rmse, 'R2 Score': r2})

results_df = pd.DataFrame(results)
print(results_df)

# ---------------------------------------------------------------------------
# 9. Hyperparameter tuning for Random Forest
# ---------------------------------------------------------------------------
param_grid = {
    'n_estimators': [100, 200, 300, 500],
    'max_depth': [10, 20, 30, 40, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2', None],
    'bootstrap': [True, False]
}

rf = RandomForestRegressor(random_state=3)
random_search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_grid,
    n_iter=100,
    cv=3,
    verbose=2,
    random_state=3,
    n_jobs=-1
)

random_search.fit(X_train, Y_train)

best_rf_model = random_search.best_estimator_
Y_pred_tuned = best_rf_model.predict(X_test)
r2_tuned = r2_score(Y_test, Y_pred_tuned)
print("Best parameters:", random_search.best_params_)
print("Tuned Random Forest R2 Score:", r2_tuned)

# ---------------------------------------------------------------------------
# 10. Save and reload the model
# ---------------------------------------------------------------------------
filename = 'best_random_forest_model.pkl'
with open(filename, 'wb') as f:
    pickle.dump(best_rf_model, f)
print(f"Model saved as {filename}")

with open(filename, 'rb') as f:
    loaded_model = pickle.load(f)

sample_data = X_test.iloc[0].values.reshape(1, -1)
predicted_price = loaded_model.predict(sample_data)
print(f"Predicted selling price: {predicted_price[0]:.2f}")
```

```python
from flask import Flask, render_template, request
import pickle
import numpy as np
import pandas as pd

# Load the trained model
model = pickle.load(open('best_model.pkl', 'rb'))

app = Flask(__name__)

# NOTE: This must match the exact column order used when training the model.
# Check your training notebook (X_train.columns) and adjust this list if needed.
FEATURE_COLUMNS = [
    'vehicle_age', 'km_driven', 'mileage', 'engine', 'max_power', 'seats',
    'seller_type_Dealer', 'seller_type_Individual', 'seller_type_Trustmark Dealer',
    'fuel_type_CNG', 'fuel_type_Diesel', 'fuel_type_Electric', 'fuel_type_LPG', 'fuel_type_Petrol',
    'transmission_type_Automatic', 'transmission_type_Manual'
]


@app.route('/')
def home():
    return render_template('index.html')


@app.route('/predict', methods=['POST'])
def predict():
    try:
        form_data = request.form

        seller_type = form_data['seller_type']
        fuel_type = form_data['fuel_type']
        transmission_type = form_data['transmission_type']

        # Dummy variables for seller_type
        seller_type_dummies = {'Dealer': 0, 'Individual': 0, 'Trustmark Dealer': 0}
        seller_type_dummies[seller_type] = 1

        # Dummy variables for fuel_type
        fuel_type_dummies = {'CNG': 0, 'Diesel': 0, 'Electric': 0, 'LPG': 0, 'Petrol': 0}
        fuel_type_dummies[fuel_type] = 1

        # Dummy variables for transmission_type
        transmission_type_dummies = {'Automatic': 0, 'Manual': 0}
        transmission_type_dummies[transmission_type] = 1

        # Build a single-row DataFrame with named columns (order-safe)
        row = {
            'vehicle_age': int(form_data['vehicle_age']),
            'km_driven': int(form_data['km_driven']),
            'mileage': float(form_data['mileage']),
            'engine': int(form_data['engine']),
            'max_power': float(form_data['max_power']),
            'seats': int(form_data['seats']),
            'seller_type_Dealer': seller_type_dummies['Dealer'],
            'seller_type_Individual': seller_type_dummies['Individual'],
            'seller_type_Trustmark Dealer': seller_type_dummies['Trustmark Dealer'],
            'fuel_type_CNG': fuel_type_dummies['CNG'],
            'fuel_type_Diesel': fuel_type_dummies['Diesel'],
            'fuel_type_Electric': fuel_type_dummies['Electric'],
            'fuel_type_LPG': fuel_type_dummies['LPG'],
            'fuel_type_Petrol': fuel_type_dummies['Petrol'],
            'transmission_type_Automatic': transmission_type_dummies['Automatic'],
            'transmission_type_Manual': transmission_type_dummies['Manual'],
        }

        input_df = pd.DataFrame([row], columns=FEATURE_COLUMNS)

        # Predict the price using the model
        predicted_price = model.predict(input_df)[0]

        # Render the result on the webpage
        return render_template(
            'index.html',
            prediction_text=f'Predicted Selling Price: ₹{round(predicted_price, 2)}'
        )

    except KeyError as e:
        return render_template('index.html', prediction_text=f'Missing form field: {e}')
    except ValueError as e:
        return render_template('index.html', prediction_text=f'Invalid input value: {e}')
    except Exception as e:
        return render_template('index.html', prediction_text=f'Error: {e}')


if __name__ == "__main__":
    app.run(debug=True, host="localhost", port=8001)
```
